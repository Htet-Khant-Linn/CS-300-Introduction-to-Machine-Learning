```python
from sklearn.datasets import load_iris
X,y=load_iris(as_frame=True,return_X_y=True)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,shuffle=True,random_state=42)
```

```python
from sklearn.model_selection import KFold,cross_val_score
kfold=KFold(n_splits=5,shuffle=True,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
lrg=LogisticRegression()
```

```python
cv_score=cross_val_score(lrg,X_train,y_train,cv=kfold,scoring='accuracy')
cv_score
```

```python
mean_score=sum(cv_score)/len(cv_score)
mean_score
```

```python
from sklearn.metrics import accuracy_score
import numpy as np
lrg.fit(X_train,y_train)
y_hat=lrg.predict(X_test)
test_acc_score=np.round(accuracy_score(y_hat,y_test),2)
print(f"The accuracy score of testing phase is {test_acc_score*100}")
```

```python
y_hat=lrg.predict(X_train)
train_acc_score=np.round(accuracy_score(y_hat,y_train),2)
print(f"The accuracy score of training phase is {train_acc_score*100}")
```
