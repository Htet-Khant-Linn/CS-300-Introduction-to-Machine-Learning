```python
from sklearn.preprocessing import OrdinalEncoder

orders=[['Iris-virginica','Iris-setosa','Iris-versicolor']]

ohe= OrdinalEncoder(categories=orders)
ohe
```

```python
from sklearn.preprocessing import OneHotEncoder
enc = OneHotEncoder()
data=enc.fit_transform(df[['Species']]).toarray()
data
```

```python
encoded_df = pd.DataFrame(data, columns=ohe.get_feature_names_out(['Species']), index=df.index)
encoded_df
```

```python
final_df = pd.concat([df.drop(columns=['Species']), encoded_df], axis=1)
final_df.head()
```
