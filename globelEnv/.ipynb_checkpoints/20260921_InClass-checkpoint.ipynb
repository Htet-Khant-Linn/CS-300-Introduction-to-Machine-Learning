```python
import pandas as pd
import seaborn as sns
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, OrdinalEncoder 
df=pd.read_csv(r".\datasets_globelEnv\Iris.csv")
# iris_df = pd.read_csv(r".\datasets_globelEnv\Iris.csv")
df
```

```python
lbl = LabelEncoder()
onehot = OneHotEncoder()
ordinalend = OrdinalEncoder()
```

```python
df['lbl_encoder']=lbl.fit_transform(df[["Species"]])
df.head()
```

```python
df['or_encoder']=ordinalend.fit_transform(df[["Species"]])
df.head()
```

```python
df['onehot_encoder']=list(onehot.fit_transform(df[["Species"]]).toarray())
df.head()
```

```python
df.Species.unique()
```

```python
pairs={
    'Iris-setosa':1.2,
    'Iris-versicolor':2.3,
    'Iris-virginica':3.5
}
pairs
```

```python
df['custom_encoder']=df['Species'].map(pairs)
df.head()
```
