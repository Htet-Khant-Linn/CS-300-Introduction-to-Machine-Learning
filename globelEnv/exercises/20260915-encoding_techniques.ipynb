```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import OneHotEncoder
```

```python
df = pd.read_csv(r".\datasets_globelEnv\Iris.csv")
df
```

# Label Encoder

```python
le = LabelEncoder()
df['LabelEncoder'] = le.fit_transform(df["Species"])    # 1D Array Expected
```

```python
df.head()
```

```python
df.Species.value_counts()
```

# Ordinal Encoder

```python
custom_order = [["Iris-setosa", "Iris-versicolor", "Iris-virginica"]]

oe = OrdinalEncoder(categories=custom_order)
df['OrdinalEncoder'] = oe.fit_transform(df[["Species"]])
```

```python
df
```

# OneHotEncoder

```python
ohe = OneHotEncoder()
ohe_encoded = ohe.fit_transform(df[["Species"]])
```

```python
print(ohe_encoded)
```

```python
type(ohe_encoded)
```

```python
ohe_array = ohe_encoded.toarray()
ohe_array
```

```python
feature_names = ohe.get_feature_names_out(['Species'])
feature_names
```

```python
ohe_df = pd.DataFrame(ohe_array, columns=feature_names)
ohe_df
```

```python
final_df = pd.concat([df, ohe_df], axis=1)
final_df
```

```python
final_df['combine'] = final_df.apply(lambda x:str(x['Species_Iris-setosa']) + ', ' + str(x['Species_Iris-versicolor']) + ', ' + str(x['Species_Iris-virginica']), axis = 1)
```
