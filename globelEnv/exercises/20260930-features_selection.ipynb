# Feature Selection and Extraction

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('../datasets_globelEnv/Iris.csv')

df.head()
```

```python
X=df.drop(columns=['Id','Species']) #df.iloc[:,1:5] df.loc
y=df['Species']   #df.iloc[:,-1]
```

# Feature Selection

```python
from sklearn.preprocessing import LabelEncoder
lbl=LabelEncoder()
df['Species']=lbl.fit_transform(df['Species'].values.reshape(-1,1))
y=df['Species']
```

```python
from sklearn.feature_selection import SelectKBest
from sklearn.feature_selection import f_classif
X_new = SelectKBest(f_classif, k=3)
X_new=X_new.fit_transform(X,y)
X_new
```

```python
from sklearn.datasets import load_diabetes

dia_X, dia_y = load_diabetes(as_frame=True, return_X_y=True)
dia_X.head()
```
