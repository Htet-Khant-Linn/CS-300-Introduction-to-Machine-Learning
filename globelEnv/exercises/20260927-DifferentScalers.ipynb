# StandardScaler, MinMaxScaler, and MaxAbsScaler

```python
import pandas as pd

iris = pd.read_csv('../datasets_globelEnv/Iris.csv')

iris.head()
```

# Standard Scaler

```python
from sklearn.preprocessing import StandardScaler
sscaler = StandardScaler()

iris['Std_Scaled_SepalLengthCM'] = sscaler.fit_transform(iris[['SepalLengthCm']])

# Inverse Transform
iris['Std_Inverse_SepalLengthCM'] = sscaler.inverse_transform(iris[['Std_Scaled_SepalLengthCM']])

iris
```

```python
print(sscaler.mean_)    # Original Mean
print(sscaler.var_)     # Original Var
print(sscaler.scale_)   # Original Std
```

```python
print(iris['Std_Scaled_SepalLengthCM'].mean())      # approx. zero
print(iris['Std_Scaled_SepalLengthCM'].std(ddof=0)) # 1
```

# MinMaxScaler

```python
from sklearn.preprocessing import MinMaxScaler

mmscaler = MinMaxScaler(feature_range=(0,1))

iris['MM_Scaled_SepalLengthCM'] = mmscaler.fit_transform(iris[['SepalLengthCm']])
iris['MM_Inverse_SepallengthCM'] = mmscaler.inverse_transform(iris[['MM_Scaled_SepalLengthCM']])
iris.head()
```

```python
print(mmscaler.data_max_)
print(mmscaler.data_min_)
print(mmscaler.data_range_)
```

```python
print(iris['MM_Scaled_SepalLengthCM'].min())
print(iris['MM_Scaled_SepalLengthCM'].max())
```

# MaxAbsScaler

```python
from sklearn.preprocessing import MaxAbsScaler

mascaler = MaxAbsScaler()

iris['MA_Scaled_SepalLengthCM'] = mascaler.fit_transform(iris[['SepalLengthCm']])
iris['MA_Inverse_SepallengthCM'] = mascaler.inverse_transform(iris[['MA_Scaled_SepalLengthCM']])
iris.head()
```

```python
print(mascaler.max_abs_)    # largest absolute value found - Speal Lenght
print(mascaler.scale_)      # actual scaling factor used to transform the feature
```

```python
print(iris['MA_Scaled_SepalLengthCM'].min())
print(iris['MA_Scaled_SepalLengthCM'].max())
```
