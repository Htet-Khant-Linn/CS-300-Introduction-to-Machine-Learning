# StandardScaler, MinMaxScaler, and MaxAbsScaler

```python
import pandas as pd

iris = pd.read_csv('../datasets_globelEnv/Iris.csv')

iris.head()
```

# Standard Scaler

```python
from sklearn.preprocessing import StandardScaler
sscaler = StandardScaler()

iris['Std_Scaled_SepalLengthCM'] = sscaler.fit_transform(iris[['SepalLengthCm']])
iris
```

```python
print(sscaler.mean_)
print(sscaler.var_)
print(sscaler.scale_)
```

```python
# Inverse Transform
iris['Std_Invert_SepalLengthCM'] = sscaler.inverse_transform(iris[['Std_Scaled_SepalLengthCM']])

iris
```
