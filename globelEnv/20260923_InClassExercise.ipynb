```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
```

```python
ames = pd.read_csv('datasets_globelEnv/AmesHousing.csv')
ames.head()
```

```python
ames.info()
```

```python
missing_values = ames.isnull().sum()
```

```python
ames.select_dtypes('float').isnull().sum()
```

```python
ames[ames['Lot Frontage'].isnull()]['Lot Frontage'].head()
```

```python
imp = SimpleImputer(missing_values=np.nan, strategy='mean')
```

```python
imp.fit_transform(ames[['Lot Frontage']])
```

```python
ames['Lot Frontage'].isnull().sum()
```

```python
ames.select_dtypes('str').isnull().sum()
```

```python
impCat = SimpleImputer(missing_values=np.nan, strategy='most_frequent')
```

```python
impCat.fit_transform(ames[['Garage Type']])
```
